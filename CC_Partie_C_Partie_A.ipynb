```python
# Sources réelles intégrées à scikit-learn : trois problèmes différents
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes, load_breast_cancer, load_wine
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, f1_score, roc_auc_score, silhouette_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
GRAINE = 42

jeu_regression = load_diabetes()
X_reg, y_reg = jeu_regression.data, jeu_regression.target
jeu_classification = load_breast_cancer()
X_cls, y_cls = jeu_classification.data, jeu_classification.target
jeu_non_supervise = load_wine()
X_clu = StandardScaler().fit_transform(jeu_non_supervise.data)  # Exploration sans cible

Xr_ent, Xr_test, yr_ent, yr_test = train_test_split(X_reg, y_reg, test_size=.25, random_state=GRAINE)
Xc_ent, Xc_test, yc_ent, yc_test = train_test_split(X_cls, y_cls, test_size=.25, stratify=y_cls, random_state=GRAINE)
print('Diabetes:', X_reg.shape, '| Breast Cancer:', X_cls.shape, '| Wine:', X_clu.shape)
print('Ces données sont réelles ; les étiquettes du vin ne sont pas employées dans le clustering.')
```

## 1. Régression linéaire

**Variante C — question spécifique :** Quel diagnostic du résidu indique une mauvaise spécification linéaire ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.linear_model import LinearRegression
modele = LinearRegression().fit(Xr_ent, yr_ent)
predictions = modele.predict(Xr_test)
print('MAE régression linéaire :', round(mean_absolute_error(yr_test, predictions), 2))
print('R² :', round(r2_score(yr_test, predictions), 3))
```

```python
# C4 — Test de sensibilité (hyperparamètre modifié ; résultats produits par le code)
import numpy as np, matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score
m = LinearRegression().fit(Xr_ent, yr_ent)
p = m.predict(Xr_test); res = yr_test - p
print('MAE test :', round(mean_absolute_error(yr_test, p), 2), '| R2 test :', round(r2_score(yr_test, p), 3),
      '| R2 entraînement :', round(r2_score(yr_ent, m.predict(Xr_ent)), 3))
print('corr(résidu, prédiction) :', round(np.corrcoef(p, res)[0, 1], 3), '| corr(|résidu|, prédiction) :', round(np.corrcoef(p, abs(res))[0, 1], 3))
# Hyperparamètre modifié : fit_intercept=False
m0 = LinearRegression(fit_intercept=False).fit(Xr_ent, yr_ent)
print('Sans constante -> MAE :', round(mean_absolute_error(yr_test, m0.predict(Xr_test)), 2), '| R2 :', round(r2_score(yr_test, m0.predict(Xr_test)), 3))
plt.scatter(p, res, s=14); plt.axhline(0, c='k'); plt.xlabel('Prédictions'); plt.ylabel('Résidus'); plt.title('Résidus vs prédictions'); plt.show()
```

#### C4 — Interprétation
**Réponse à la question :** une mauvaise spécification linéaire se voit dans le graphique *résidus vs valeurs prédites* : une **courbure** (U ou en S) signale une non-linéarité omise, un **entonnoir** signale une hétéroscédasticité ; le QQ-plot des résidus et leur dépendance avec une variable absente sont des diagnostics complémentaires. Ici la corrélation résidu/prédiction est faible (−0,078, pas de tendance linéaire résiduelle) mais corr(|résidu|, prédiction) = 0,286 : l'erreur grandit avec le niveau prédit, signe d'**hétéroscédasticité modérée**.

**Sensibilité :** avec `fit_intercept=False` le modèle s'effondre (MAE 156,36 ; R² −3,94 contre MAE 41,55 ; R² 0,485) car la cible a une moyenne élevée (~150) alors que les variables sont centrées : la constante est indispensable. R² entraînement 0,519 vs test 0,485 : faible écart, pas de surapprentissage mais un modèle limité.

**Ce que l'on ne peut pas conclure :** une seule séparation (111 observations de test) ne donne pas d'intervalle de confiance ; R² ≈ 0,49 ne prouve pas que la relation soit « vraiment » linéaire, ni causale.

## 2. Ridge

**Variante C — question spécifique :** Comment évoluent les coefficients quand alpha augmente ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.linear_model import Ridge
modele = make_pipeline(StandardScaler(), Ridge(alpha=12.0)).fit(Xr_ent, yr_ent)
print('MAE Ridge :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

```python
# C4 — Test de sensibilité (hyperparamètre modifié ; résultats produits par le code)
from sklearn.linear_model import Ridge
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, r2_score
for a in [0.01, 1, 12, 100, 1000]:
    m = make_pipeline(StandardScaler(), Ridge(alpha=a)).fit(Xr_ent, yr_ent); c = m[-1].coef_
    print(f'alpha={a:>7} | MAE={mean_absolute_error(yr_test, m.predict(Xr_test)):.2f} | R2={r2_score(yr_test, m.predict(Xr_test)):.3f} | norme coef={np.linalg.norm(c):.2f} | max|coef|={abs(c).max():.2f}')
```

#### C4 — Interprétation
**Réponse :** quand $\alpha$ augmente, les coefficients **rétrécissent de façon continue vers zéro sans jamais s'annuler** : norme $\ell_2$ 70,31 ($\alpha=0{,}01$) → 59,75 (1) → 41,99 (12) → 33,79 (100) → 15,16 (1000) ; le plus grand coefficient passe de 44,01 à 8,63.

**Effet sur l'erreur :** MAE 41,55 → 41,51 → 41,46 ($\alpha=12$) → 41,93 (100) → **50,44** (1000, R² 0,356) : un léger rétrécissement stabilise, un rétrécissement excessif crée du **sous-apprentissage** (biais). Les différences entre $\alpha=0{,}01$ et 12 (0,09 point de MAE) sont négligeables.

**Limites :** pas de conclusion sur le « meilleur » $\alpha$ à partir du test seul : il se choisit par validation croisée sur l'entraînement ; une seule séparation et une seule graine ne mesurent pas l'incertitude.

```python
# C4 — Test de sensibilité (hyperparamètre modifié ; résultats produits par le code)
from sklearn.linear_model import Lasso
for a in [0.1, 2, 10]:
    m = make_pipeline(StandardScaler(), Lasso(alpha=a, max_iter=15000)).fit(Xr_ent, yr_ent)
    nz = np.count_nonzero(m[-1].coef_)
    print(f'alpha={a:>4} | non nuls={nz} | annulés={Xr_ent.shape[1]-nz} | MAE={mean_absolute_error(yr_test, m.predict(Xr_test)):.2f} | R2={r2_score(yr_test, m.predict(Xr_test)):.3f}')
```

#### C4 — Interprétation
**Réponse (10 variables) :** $\alpha=0{,}1$ → **1** variable annulée (9 non nulles) ; $\alpha=2$ → **2** annulées (8 non nulles) ; $\alpha=10$ → **6** annulées (4 non nulles). Plus $\alpha$ est grand, plus le modèle est parcimonieux.

**Interprétation :** MAE 41,48 → 41,37 → 43,64. Jusqu'à $\alpha=2$ on élimine des variables sans perdre en précision (R² 0,502 à $\alpha=2$) ; à $\alpha=10$ l'élimination devient trop agressive et l'erreur augmente (compromis parcimonie/précision).

**Limites :** les variables retenues dépendent de l'échantillon et de la corrélation entre variables ; une sélection ne signifie pas une causalité. Une seule séparation ne permet pas de juger la stabilité de la sélection.

## 4. Régression polynomiale

**Variante C — question spécifique :** Le degré 2 améliore-t-il le test ou seulement l’entraînement ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele = make_pipeline(StandardScaler(), PolynomialFeatures(degree=2, include_bias=False), Ridge(alpha=100.)).fit(Xr_ent, yr_ent)
print('R² polynomial :', round(r2_score(yr_test, modele.predict(Xr_test)), 3))
```

```python
# C4 — Test de sensibilité (hyperparamètre modifié ; résultats produits par le code)
from sklearn.preprocessing import PolynomialFeatures
for d, a in [(1, 100), (2, 100), (2, 1), (3, 100), (3, 1)]:
    m = make_pipeline(StandardScaler(), PolynomialFeatures(degree=d, include_bias=False), Ridge(alpha=a)).fit(Xr_ent, yr_ent)
    print(f'degré={d} alpha={a:>3} | R2 entraînement={r2_score(yr_ent, m.predict(Xr_ent)):.3f} | R2 test={r2_score(yr_test, m.predict(Xr_test)):.3f}')
```

#### C4 — Interprétation
**Réponse :** avec le réglage fourni (degré 2, $\alpha=100$) le R² d'entraînement monte de 0,502 (degré 1) à 0,557 mais le R² de test ne passe que de 0,490 à **0,501** : l'amélioration est **surtout sur l'entraînement** ; le gain test (+0,011 sur 111 observations) est trop faible pour être jugé réel.

**Sensibilité :** si on relâche la régularisation ($\alpha=1$), le test se dégrade (0,455) alors que l'entraînement monte (0,600) ; au degré 3 : 0,662/0,368 ($\alpha=100$) puis 0,808/**−0,558** ($\alpha=1$) → surajustement massif (R² négatif = pire que prédire la moyenne).

**Limites :** une seule séparation ; l'écart d'entraînement/test, et non le R² d'entraînement, mesure la généralisation.

## 5. Arbre de régression

**Variante C — question spécifique :** Quel effet de max_depth sur le biais et la variance ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.tree import DecisionTreeRegressor
modele = DecisionTreeRegressor(max_depth=3, min_samples_leaf=8, random_state=GRAINE).fit(Xr_ent, yr_ent)
print('MAE arbre :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

```python
# C4 — Test de sensibilité (hyperparamètre modifié ; résultats produits par le code)
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import cross_val_score, KFold
cv = KFold(5, shuffle=True, random_state=GRAINE)
for d in [1, 2, 3, 5, 8, None]:
    m = DecisionTreeRegressor(max_depth=d, min_samples_leaf=8, random_state=GRAINE).fit(Xr_ent, yr_ent)
    cvmae = -cross_val_score(DecisionTreeRegressor(max_depth=d, min_samples_leaf=8, random_state=GRAINE), Xr_ent, yr_ent, cv=cv, scoring='neg_mean_absolute_error').mean()
    print(f'max_depth={str(d):>4} | MAE entraînement={mean_absolute_error(yr_ent, m.predict(Xr_ent)):.2f} | MAE test={mean_absolute_error(yr_test, m.predict(Xr_test)):.2f} | MAE validation croisée={cvmae:.2f}')
```

#### C4 — Interprétation
**Réponse :** `max_depth` règle le compromis **biais/variance**. Profondeur faible → **biais élevé** (sous-apprentissage) : MAE 55,07 (entraînement) et 55,10 (test) à la profondeur 1. Profondeur élevée → **variance élevée** : l'erreur d'entraînement baisse (36,71 à 5, 34,15 à 8, 33,63 sans limite) pendant que l'erreur de test remonte (47,01, 50,45, 50,27). Le fort écart entraînement/test à partir de la profondeur 5 est la signature du surajustement.

**Choix du réglage :** il se fait par validation croisée sur l'entraînement (pas sur le test) ; le tableau affiche la MAE par validation croisée pour chaque profondeur, et le minimum doit être lu dans cette colonne (profondeurs 2 à 3 environ).

**Limites :** un arbre est instable (un petit changement des données peut changer la structure) ; une seule graine et un seul découpage ne le montrent pas.

## 6. Forêt aléatoire — régression

**Variante C — question spécifique :** La forêt réduit-elle la variance par rapport à l’arbre ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.ensemble import RandomForestRegressor
modele = RandomForestRegressor(n_estimators=60, max_depth=6, n_jobs=-1, random_state=GRAINE).fit(Xr_ent, yr_ent)
print('MAE forêt :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

```python
# C4 — Test de sensibilité (hyperparamètre modifié ; résultats produits par le code)
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
arbre = DecisionTreeRegressor(max_depth=6, min_samples_leaf=8, random_state=GRAINE).fit(Xr_ent, yr_ent)
print('Arbre profondeur 6 -> MAE test :', round(mean_absolute_error(yr_test, arbre.predict(Xr_test)), 2))
for n in [1, 10, 60, 300]:
    m = RandomForestRegressor(n_estimators=n, max_depth=6, n_jobs=-1, random_state=GRAINE).fit(Xr_ent, yr_ent)
    print(f'n_estimators={n:>3} | MAE test={mean_absolute_error(yr_test, m.predict(Xr_test)):.2f} | R2={r2_score(yr_test, m.predict(Xr_test)):.3f}')
for n in [10, 300]:
    s = [mean_absolute_error(yr_test, RandomForestRegressor(n_estimators=n, max_depth=6, random_state=k).fit(Xr_ent, yr_ent).predict(Xr_test)) for k in range(5)]
    print(f'n_estimators={n:>3} sur 5 graines -> MAE moyenne={np.mean(s):.2f} | écart-type={np.std(s):.3f}')
```

#### C4 — Interprétation
**Réponse : oui.** À profondeur égale (6), l'arbre seul obtient une MAE de test de **49,37** (R² 0,321), la forêt de 60 arbres **43,11** (R² 0,460) et celle de 300 arbres **42,53** (R² 0,486).

**Sensibilité à `n_estimators` :** 1 arbre → 53,37 (R² 0,080) ; 10 → 44,67 ; 60 → 43,11 ; 300 → 42,53 : le gain est fort au début puis **plafonne**. Sur 5 graines, l'écart-type de la MAE passe de **0,813** (10 arbres) à **0,298** (300 arbres) : plus d'arbres → prédiction plus **stable** (variance liée au hasard réduite).

**Limites :** la forêt réduit la variance, pas le biais ; elle ne corrige pas un modèle mal spécifié. Les 5 graines ne portent que sur l'aléa du modèle, pas sur l'aléa du découpage entraînement/test.

## 7. Gradient Boosting — régression

**Variante C — question spécifique :** Que change max_iter sur le coût et le risque de surajustement ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

## 26. Epsilon-greedy

**Variante C — question spécifique :** Quel effet de epsilon sur la concentration des choix ?

```python
# Exploration epsilon-greedy sur le même bandit simulé
rng = np.random.default_rng(GRAINE)
probabilites_simulees = np.array([.35, .55, .65, .45])
recompenses = np.zeros(4); nombre_tirages = np.zeros(4)
epsilon = .2
for etape in range(1500):
    bras = int(rng.integers(4)) if rng.random() < epsilon or not np.all(nombre_tirages) else int(np.argmax(recompenses / np.maximum(nombre_tirages, 1)))
    gain = float(rng.random() < probabilites_simulees[bras])
    recompenses[bras] += gain; nombre_tirages[bras] += 1
print('Fréquences de sélection :', nombre_tirages.astype(int))
print('Valeurs estimées :', np.round(recompenses / np.maximum(nombre_tirages, 1), 3))
```

```python
# C4 — Test de sensibilité (hyperparamètre modifié ; résultats produits par le code)
def egreedy(eps, n=1500, graine=GRAINE):
    rng = np.random.default_rng(graine); r = np.zeros(4); c = np.zeros(4); tot = 0.
    for e in range(n):
        b = int(rng.integers(4)) if rng.random() < eps or not np.all(c) else int(np.argmax(r / np.maximum(c, 1)))
        g = float(rng.random() < p_vrai[b]); r[b] += g; c[b] += 1; tot += g
    return c.astype(int), tot
for eps in [0, 0.05, 0.2, 0.5, 1]:
    c, tot = egreedy(eps)
    res = [egreedy(eps, graine=s) for s in range(20)]
    print(f'eps={eps:<4} | graine 42 : tirages={c}, gain={tot:.0f} | 20 graines : gain moyen={np.mean([x[1] for x in res]):.1f} (écart-type {np.std([x[1] for x in res]):.1f}), part du meilleur bras={np.mean([x[0][2]/1500 for x in res]):.3f}')
```

#### C4 — Interprétation
**Réponse :** $\varepsilon$ détermine la **concentration** des choix. À $\varepsilon=0{,}2$ (graine 42), 1215 des 1500 tirages vont au bras 2 (gain 911) ; à 0,5 : 908 tirages (gain 867) ; à 1 (aléatoire pur) : 351 (gain 750). Plus $\varepsilon$ est grand, plus les choix sont dispersés.

**Piège de $\varepsilon$ trop petit :** avec $\varepsilon=0$ (et 0,05) la graine 42 s'enferme sur le **mauvais bras 1** (1496 puis 1428 tirages ; gain 822 et 832). Sur 20 graines, la part moyenne du bras 2 est 0,597 ($\varepsilon=0$), **0,838** (0,05), 0,786 (0,2), 0,585 (0,5), 0,248 (1) ; le gain moyen est 865,4 (écart-type 148,7), **937,7** (45,8), 914,5 (25,5), 856,8, 740,5. À $\varepsilon=0$, le meilleur bras est manqué dans 8 graines sur 20.

**Lecture :** un compromis intermédiaire (0,05 à 0,2 sur cet horizon) est préférable ; $\varepsilon$ constant garde un coût d'exploration permanent.

**Limites :** bandit **simulé**, 1500 tirages, 4 bras ; le meilleur $\varepsilon$ dépend de l'horizon et des écarts entre bras.

## 27. Q-learning

**Variante C — question spécifique :** Étudiez les récompenses des épisodes, pas seulement Q moyen.

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

```python
# Compat. Gymnasium récent : CliffWalking-v0 est déprécié -> v1 (même environnement)
# Q-learning manuel — environnement CliffWalking (Gymnasium)
import gymnasium as gym
env = gym.make('CliffWalking-v1')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    for pas in range(120):
        action = int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
        suivant, recompense, termine, tronque, _ = env.step(action)
        cible = recompense + (0 if termine else .95 * np.max(q_table[suivant]))
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat = suivant
        if termine or tronque: break
print('Table Q :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
env.close()
```

```python
# Compat. Gymnasium récent : CliffWalking-v0 est déprécié -> v1 (même environnement)
# SARSA manuel — même environnement, politique epsilon-greedy suivie
import gymnasium as gym
env = gym.make('CliffWalking-v1')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
def choisir_action(etat, epsilon):
    return int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    action = choisir_action(etat, epsilon)
    for pas in range(120):
        suivant, recompense, termine, tronque, _ = env.step(action)
        action_suivante = choisir_action(suivant, epsilon)
        cible = recompense + (0 if termine else .95 * q_table[suivant, action_suivante])
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat, action = suivant, action_suivante
        if termine or tronque: break
print('Table Q SARSA :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
env.close()
```

## 29. DQN — cible conceptuelle

**Variante C — question spécifique :** Pourquoi la cible est-elle égale à la récompense en état terminal ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

```python
# Cibles DQN : démonstration du calcul de Bellman, SANS réseau neuronal
recompenses = np.array([-1., -100., 0.])
termine = np.array([False, True, False])
meilleure_valeur_suivante = np.array([-3., -2., 5.])
gamma = .9
cibles = recompenses + gamma * (~termine) * meilleure_valeur_suivante
print('Cibles de Bellman calculées :', np.round(cibles, 2))
print('Ce calcul ne constitue pas un entraînement DQN.')
```

```python
# C4 — Test de sensibilité (hyperparamètre modifié ; résultats produits par le code)
recompenses = np.array([-1., -100., 0.]); termine = np.array([False, True, False]); meilleure_suivante = np.array([-3., -2., 5.])
for gamma in [0, 0.5, 0.9, 0.99]:
    print(f'gamma={gamma:<4} | cibles de Bellman={np.round(recompenses + gamma * (~termine) * meilleure_suivante, 2)}')
print('Ce calcul ne constitue pas un entraînement DQN.')
```

#### C4 — Interprétation
**Réponse :** à l'état terminal, il n'existe **aucun état suivant** : la valeur future est nulle par définition (l'épisode s'arrête). La cible se réduit à $y=r$ ; ici −100 pour la transition terminale (chute), quelle que soit la valeur « estimée » −2 du réseau pour $s'$. Garder le terme $\gamma\max Q(s',a')$ ajouterait de la valeur fictive.

**Sensibilité à $\gamma$ :** $\gamma=0$ → cibles [−1 ; −100 ; 0] (agent myope) ; 0,5 → [−2,5 ; −100 ; 2,5] ; **0,9** → [**−3,7 ; −100 ; 4,5**] (valeur fournie) ; 0,99 → [−3,97 ; −100 ; 4,95]. La cible terminale (−100) ne dépend jamais de $\gamma$ ; les autres incorporent d'autant plus le futur que $\gamma$ est proche de 1.

**Limites :** cette cellule **n'entraîne pas de DQN** (pas de réseau, pas de rejeu, pas d'optimisation) ; les valeurs `meilleure_valeur_suivante` sont choisies à la main, pas estimées. Aucune conclusion sur la performance d'un DQN n'est possible.
